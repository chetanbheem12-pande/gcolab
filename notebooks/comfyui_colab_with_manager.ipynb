### Run ComfyUI with cloudflared (Recommended Way)




```python
!wget -P ~ https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64.deb
!dpkg -i ~/cloudflared-linux-amd64.deb

import subprocess
import threading
import time
import socket
import urllib.request

def iframe_thread(port):
  while True:
      time.sleep(0.5)
      sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
      result = sock.connect_ex(('127.0.0.1', port))
      if result == 0:
        break
      sock.close()
  print("\nComfyUI finished loading, trying to launch cloudflared (if it gets stuck here cloudflared is having issues)\n")

  p = subprocess.Popen(["cloudflared", "tunnel", "--url", "http://127.0.0.1:{}".format(port)], stdout=subprocess.PIPE, stderr=subprocess.PIPE)
  for line in p.stderr:
    l = line.decode()
    if "trycloudflare.com " in l:
      print("This is the URL to access ComfyUI:", l[l.find("http"):], end='')
    #print(l, end='')


threading.Thread(target=iframe_thread, daemon=True, args=(8188,)).start()

!python main.py --dont-print-server
```

```python
# Cell 3 — Launch ComfyUI (CPU) with comfy_aimdo shim and ngrok (or fallback to cloudflared)
from pathlib import Path
import os, sys, subprocess, time, re, socket, urllib.request
from datetime import datetime, timedelta
from getpass import getpass

# ---------- config ----------
WORKSPACE = "/content/drive/MyDrive/ComfyUI"
CUSTOM_NODES = os.path.join(WORKSPACE, "custom_nodes")
MODEL_YAML = os.path.join(WORKSPACE, "models_paths.yaml")
COMFYUI_PORT = 8188
COMFYUI_READY_TIMEOUT = 180
TUNNEL_START_TIMEOUT = 40

os.makedirs(CUSTOM_NODES, exist_ok=True)
os.chdir(WORKSPACE)

# ---------- install system ffmpeg (video support) ----------
!apt-get update -qq
!apt-get install -y -qq ffmpeg

# ---------- comfy_aimdo shim (minimal) ----------
shim_dir = os.path.join(CUSTOM_NODES, "comfy_aimdo")
os.makedirs(shim_dir, exist_ok=True)
# __init__
with open(os.path.join(shim_dir, "__init__.py"), "w", encoding="utf-8") as f:
    f.write("# Auto shim for comfy_aimdo (placeholder)\n")
# control.py
with open(os.path.join(shim_dir, "control.py"), "w", encoding="utf-8") as f:
    f.write("def register_control(*a, **k):\n    return None\n")
# torch.py shim
with open(os.path.join(shim_dir, "torch.py"), "w", encoding="utf-8") as f:
    f.write("try:\n    import torch as _real_torch\nexcept Exception:\n    _real_torch = None\n\ndef real_available():\n    return _real_torch is not None\n")
# small stubs for other missing submodules (add more if ComfyUI complains)
for name in ("model_vbar", "model_utils", "vbar_api", "control_api"):
    p = os.path.join(shim_dir, f"{name}.py")
    if not os.path.exists(p):
        with open(p, "w", encoding="utf-8") as f:
            f.write(f"# shim for comfy_aimdo.{name}\ndef missing(*a, **k):\n    raise RuntimeError('shim: comfy_aimdo.{name} not installed')\n")

if CUSTOM_NODES not in sys.path:
    sys.path.insert(0, CUSTOM_NODES)
os.environ["PYTHONPATH"] = f"{CUSTOM_NODES}:{os.environ.get('PYTHONPATH','')}"

print("✅ comfy_aimdo shim ensured at:", shim_dir)

# ---------- write models_paths.yaml if missing ----------
from pathlib import Path
model_dir = os.path.join(WORKSPACE, "models")
os.makedirs(model_dir, exist_ok=True)
folders = [p for p in sorted(Path(model_dir).iterdir()) if p.is_dir()]
with open(MODEL_YAML, "w", encoding="utf-8") as yf:
    yf.write("models:\n")
    for folder in folders:
        yf.write(f"  {folder.name}: |\n")
        yf.write(f"    {str(folder.resolve())}\n")
print("✅ models YAML written:", MODEL_YAML)
for f in folders: print(" -", f)

# ---------- helpers ----------
def wait_for_port_open(host, port, timeout_s):
    deadline = datetime.now() + timedelta(seconds=timeout_s)
    while datetime.now() < deadline:
        try:
            with socket.create_connection((host, port), timeout=2):
                return True
        except Exception:
            time.sleep(0.4)
    return False

def comfyui_http_ready(url, timeout_s=3):
    try:
        with urllib.request.urlopen(url, timeout=timeout_s) as resp:
            return True
    except Exception:
        return False

# ---------- decide CPU mode and prepare args ----------
print("No GPU detected / or CUDA unavailable in this Colab — launching ComfyUI with --cpu.")
args = [sys.executable, "main.py", "--extra-model-paths-config", MODEL_YAML,
        "--input-directory", f"{WORKSPACE}/input", "--output-directory", f"{WORKSPACE}/output", "--cpu"]

# ---------- start ComfyUI (capture logs) ----------
print("\n🚀 Starting ComfyUI (logs streaming)...\n")
env = os.environ.copy()
env["PYTHONPATH"] = os.environ.get("PYTHONPATH", "")
proc = subprocess.Popen(args, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, env=env, bufsize=1)

# stream early logs
start = time.time()
while time.time() - start < 12:
    line = proc.stdout.readline()
    if not line:
        time.sleep(0.1)
        continue
    print(line, end="")

# wait for port bind
print(f"\n⏳ Waiting for ComfyUI to bind to port {COMFYUI_PORT} (timeout {COMFYUI_READY_TIMEOUT}s)...")
if not wait_for_port_open("127.0.0.1", COMFYUI_PORT, COMFYUI_READY_TIMEOUT):
    print("❌ ComfyUI failed to bind port within timeout. Tail logs:")
    for _ in range(200):
        l = proc.stdout.readline()
        if not l: break
        print(l, end="")
    proc.kill()
    raise RuntimeError("ComfyUI failed to start or bind the port. See logs above.")

# wait for HTTP ready
print("✅ Port open. Waiting for HTTP readiness...")
deadline = time.time() + COMFYUI_READY_TIMEOUT
http_ok = False
while time.time() < deadline:
    if comfyui_http_ready(f"http://127.0.0.1:{COMFYUI_PORT}", timeout_s=3):
        http_ok = True
        break
    while True:
        line = proc.stdout.readline()
        if not line: break
        print(line, end="")
    time.sleep(0.5)

if not http_ok:
    print("❌ ComfyUI not responding to HTTP. Tail logs printed above.")
    proc.kill()
    raise RuntimeError("ComfyUI HTTP readiness failed.")

print("✅ ComfyUI is ready locally at http://127.0.0.1:{}".format(COMFYUI_PORT))

# ---------- Tunnel selection: ngrok (preferred) or cloudflared fallback ----------
print("\nTunneling options:\n • ngrok (recommended) — requires an ngrok authtoken (free) \n • cloudflared quick tunnel — no account required (may be transient).")

# first check env var
ngrok_token = os.environ.get("NGROK_AUTH_TOKEN")
if not ngrok_token:
    # ask user to paste token (safe prompt)
    print("\nIf you have an ngrok authtoken, paste it now. Otherwise press Enter to use cloudflared fallback.")
    print("Get a token at: https://dashboard.ngrok.com/get-started/your-authtoken")
    try:
        ngrok_token = getpass("ngrok authtoken (paste or press Enter to skip): ")
    except Exception:
        ngrok_token = ""

# try ngrok if token provided
if ngrok_token:
    try:
        from pyngrok import ngrok, conf
        print("Registering ngrok token and starting tunnel...")
        conf.get_default().auth_token = ngrok_token
        # ensure ngrok binary is available (pyngrok downloads it automatically)
        tunnel = ngrok.connect(COMFYUI_PORT)
        public_url = tunnel.public_url
        print("\n✅ ngrok public URL:", public_url)
        # embed in Colab
        try:
            from IPython.display import HTML, display
            display(HTML(f'<iframe src="{public_url}" width="100%" height="780"></iframe>'))
        except Exception:
            pass
        print("Open the ngrok URL above to access ComfyUI.")
        raise SystemExit(0)  # done
    except Exception as e:
        print("ngrok failed:", repr(e))
        print("Falling back to cloudflared quick tunnel...")

# fallback to cloudflared quick tunnel (no account needed)
print("\nStarting cloudflared quick tunnel (fallback)...")
# ensure cloudflared present
if os.system("command -v cloudflared > /dev/null 2>&1") != 0:
    print("Installing cloudflared...")
    !wget -q https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64.deb -O /tmp/cloudflared.deb
    !dpkg -i /tmp/cloudflared.deb

# launch and parse trycloudflare URL
tproc = subprocess.Popen(["cloudflared", "tunnel", "--url", f"http://127.0.0.1:{COMFYUI_PORT}"],
                         stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)

public_url = None
t_start = time.time()
while time.time() - t_start < TUNNEL_START_TIMEOUT:
    line = tproc.stdout.readline()
    if not line:
        time.sleep(0.1); continue
    print(line, end="")
    m = re.search(r"https://[a-z0-9\-]+\.trycloudflare\.com", line)
    if m:
        public_url = m.group(0)
        break

if public_url:
    print("\n✅ Cloudflare quick-tunnel URL:", public_url)
    try:
        from IPython.display import HTML, display
        display(HTML(f'<iframe src="{public_url}" width="100%" height="780"></iframe>'))
    except Exception:
        pass
    print("Open the above URL to access ComfyUI. If you get 502, refresh in ~10-60s — sometimes quick-tunnel needs a moment.")
else:
    # print remaining logs for debugging
    print("\n⚠️ Could not find a trycloudflare URL in cloudflared output. Recent cloudflared log:")
    for _ in range(200):
        l = tproc.stdout.readline()
        if not l: break
        print(l, end="")
    raise RuntimeError("Failed to create a public tunnel with ngrok/cloudflared.")
```
