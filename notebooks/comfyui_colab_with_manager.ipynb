```python
# ==========================================
# Cell 3 — ComfyUI Setup + comfy_aimdo full shim + Cloudflare Tunnel
# (creates comfy_aimdo.control and comfy_aimdo.torch shims if needed)
# ==========================================
from pathlib import Path
import os, sys, subprocess, time, re, socket, urllib.request
from datetime import datetime, timedelta

# -------- user options --------
USE_GOOGLE_DRIVE = True
USE_COMFYUI_MANAGER = True
COMFYUI_PORT = 8188
COMFYUI_READY_TIMEOUT = 180
CLOUDFLARED_START_TIMEOUT = 60

# -------- mount drive safely --------
if USE_GOOGLE_DRIVE:
    from google.colab import drive
    drive.mount('/content/drive', force_remount=True)
    if not os.path.ismount("/content/drive"):
        raise RuntimeError("Google Drive did not mount. Reconnect and re-run the cell.")

# -------- paths --------
WORKSPACE = "/content/drive/MyDrive/ComfyUI"
CUSTOM_NODES = os.path.join(WORKSPACE, "custom_nodes")
os.makedirs(WORKSPACE, exist_ok=True)
os.makedirs(CUSTOM_NODES, exist_ok=True)
os.chdir(WORKSPACE)

# -------- clone comfyui if needed --------
if not os.path.exists(os.path.join(WORKSPACE, "main.py")):
    print("Cloning ComfyUI...")
    !git clone https://github.com/comfyanonymous/ComfyUI .

# -------- install light deps --------
!pip install -q accelerate einops transformers safetensors aiohttp pyyaml Pillow scipy tqdm psutil tokenizers
!pip install -q torchsde kornia spandrel soundfile sentencepiece

# -------- ensure ComfyUI-Manager present (optional) --------
if USE_COMFYUI_MANAGER:
    mgr_path = os.path.join(CUSTOM_NODES, "ComfyUI-Manager")
    if not os.path.exists(mgr_path):
        os.chdir(CUSTOM_NODES)
        !git clone https://github.com/ltdrdata/ComfyUI-Manager
        os.chdir(WORKSPACE)

# -------- build models YAML (ComfyUI format) --------
MODEL_DIR = os.path.join(WORKSPACE, "models")
os.makedirs(MODEL_DIR, exist_ok=True)
folders = [p for p in sorted(Path(MODEL_DIR).iterdir()) if p.is_dir()]
MODEL_YAML = os.path.join(WORKSPACE, "models_paths.yaml")
with open(MODEL_YAML, "w", encoding="utf-8") as yf:
    yf.write("models:\n")
    for folder in folders:
        key = folder.name
        path_str = os.path.abspath(str(folder))
        yf.write(f"  {key}: |\n")
        for line in path_str.splitlines():
            yf.write(f"    {line}\n")
print("✅ models YAML created:", MODEL_YAML)
for f in folders:
    print(" -", f)

# -------- ensure input/output exist --------
os.makedirs(os.path.join(WORKSPACE, "input"), exist_ok=True)
os.makedirs(os.path.join(WORKSPACE, "output"), exist_ok=True)

# -------- try to pip install the real comfy_aimdo once (optional) --------
print("\nAttempting pip install comfy_aimdo from GitHub (one-shot)...")
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "git+https://github.com/comfy-aimdo/comfy_aimdo.git"], stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)

# -------- create / ensure shim package comfy_aimdo exists (control + torch) --------
shim_dir = os.path.join(CUSTOM_NODES, "comfy_aimdo")
os.makedirs(shim_dir, exist_ok=True)

# __init__.py (keep minimal)
init_path = os.path.join(shim_dir, "__init__.py")
if not os.path.exists(init_path):
    with open(init_path, "w", encoding="utf-8") as f:
        f.write("# Local shim for comfy_aimdo (auto-generated)\n")

# control.py shim (some imports check for comfy_aimdo.control)
control_path = os.path.join(shim_dir, "control.py")
if not os.path.exists(control_path):
    with open(control_path, "w", encoding="utf-8") as f:
        f.write('''# comfy_aimdo.control shim
# Minimal stub so "import comfy_aimdo.control" succeeds.
def register_control(*args, **kwargs):
    # Placeholder - real comfy_aimdo.control provides runtime behavior.
    return None
''')

# torch.py shim (this was missing and caused the crash)
torch_shim_path = os.path.join(shim_dir, "torch.py")
if not os.path.exists(torch_shim_path):
    with open(torch_shim_path, "w", encoding="utf-8") as f:
        f.write(r'''# comfy_aimdo.torch shim
# This module tries to forward to real torch if available, otherwise provides harmless placeholders.
try:
    import torch as _real_torch
except Exception:
    _real_torch = None

# expose a few convenient names
torch = _real_torch

def real_available():
    return _real_torch is not None

# placeholder wrappers for functions ComfyUI or comfy_aimdo might try to call
def tensor(data):
    if _real_torch is not None:
        return _real_torch.tensor(data)
    raise RuntimeError("torch not available in environment (shim)")

def load_state_dict(*args, **kwargs):
    if _real_torch is not None:
        return _real_torch.load(*args, **kwargs)
    raise RuntimeError("torch not available in environment (shim)")

# If code expects attributes under comfy_aimdo.torch, they won't crash on import.
class _Missing:
    def __getattr__(self, name):
        def _missing(*args, **kwargs):
            raise RuntimeError(f"comfy_aimdo.torch shim: function {name} is not implemented. Install real comfy_aimdo or torch.")
        return _missing

# provide a fallback object to avoid attribute errors when imported
if _real_torch is None:
    fallback = _Missing()
    # populate a few common names to be safe
    fallback.tensor = tensor
    fallback.load_state_dict = load_state_dict
    # expose as module-level names
    globals().update({
        "fallback": fallback,
    })
''')

# ensure python path sees custom_nodes
if CUSTOM_NODES not in sys.path:
    sys.path.insert(0, CUSTOM_NODES)
os.environ["PYTHONPATH"] = f"{CUSTOM_NODES}:{os.environ.get('PYTHONPATH','')}"

print("\nShims ensured at:", shim_dir)

# -------- install cloudflared if needed --------
if os.system("command -v cloudflared > /dev/null 2>&1") != 0:
    print("\nInstalling cloudflared...")
    !wget -q https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64.deb -O /tmp/cloudflared.deb
    !dpkg -i /tmp/cloudflared.deb

# -------- helper readiness functions --------
def wait_for_port_open(host, port, timeout_s):
    deadline = datetime.now() + timedelta(seconds=timeout_s)
    while datetime.now() < deadline:
        try:
            with socket.create_connection((host, port), timeout=2):
                return True
        except Exception:
            time.sleep(0.4)
    return False

def comfyui_http_ready(url, timeout_s=3):
    try:
        with urllib.request.urlopen(url, timeout=timeout_s) as resp:
            return True
    except Exception:
        return False

# -------- launch ComfyUI (capture logs) --------
print("\n🚀 Launching ComfyUI (logs will appear below)...\n")
env = os.environ.copy()
env["PYTHONPATH"] = os.environ.get("PYTHONPATH", "")

proc = subprocess.Popen([sys.executable, "main.py",
                         "--extra-model-paths-config", MODEL_YAML,
                         "--input-directory", f"{WORKSPACE}/input",
                         "--output-directory", f"{WORKSPACE}/output"],
                        stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, env=env, bufsize=1)

# stream initial logs
start_t = time.time()
initial_lines = []
while time.time() - start_t < 10:
    line = proc.stdout.readline()
    if not line:
        time.sleep(0.1)
        continue
    initial_lines.append(line)
    print(line, end="")
    # if fatal import still occurs, break and show logs
    if "ModuleNotFoundError" in line or "Traceback (most recent call last)" in line:
        break

# if crash detected quickly, print and abort
if any("ModuleNotFoundError" in l for l in initial_lines):
    print("\n❌ ComfyUI crashed during startup. Showing collected logs:")
    print("".join(initial_lines[-200:]))
    proc.kill()
    raise RuntimeError("ComfyUI failed to start (see logs above).")

# wait for server port
print("\n⏳ Waiting for ComfyUI to bind to port {}...".format(COMFYUI_PORT))
if not wait_for_port_open("127.0.0.1", COMFYUI_PORT, timeout_s=COMFYUI_READY_TIMEOUT):
    print("❌ ComfyUI did not bind the port within timeout. Tail of logs below:")
    for _ in range(200):
        l = proc.stdout.readline()
        if not l:
            break
        print(l, end="")
    proc.kill()
    raise RuntimeError("ComfyUI failed to start or bind the port.")

# wait for http readiness
print("✅ Port open. Waiting for HTTP readiness...")
deadline = time.time() + COMFYUI_READY_TIMEOUT
http_ok = False
while time.time() < deadline:
    if comfyui_http_ready(f"http://127.0.0.1:{COMFYUI_PORT}", timeout_s=3):
        http_ok = True
        break
    while True:
        line = proc.stdout.readline()
        if not line:
            break
        print(line, end="")
    time.sleep(0.5)

if not http_ok:
    print("❌ ComfyUI not responding to HTTP. Tail of logs:")
    for _ in range(200):
        l = proc.stdout.readline()
        if not l:
            break
        print(l, end="")
    proc.kill()
    raise RuntimeError("ComfyUI HTTP readiness failed.")

print("✅ ComfyUI ready locally. Starting Cloudflared (capturing URL)...")

tunnel = subprocess.Popen(["cloudflared", "tunnel", "--url", f"http://127.0.0.1:{COMFYUI_PORT}"],
                          stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)

public_url = None
t_end = time.time() + CLOUDFLARED_START_TIMEOUT
while time.time() < t_end:
    line = tunnel.stdout.readline()
    if not line:
        time.sleep(0.1)
        continue
    print(line, end="")
    m = re.search(r"https://[a-z0-9\-]+\.trycloudflare\.com", line)
    if m:
        public_url = m.group(0)
        break

if public_url:
    print("\n✅ ComfyUI public URL:", public_url)
    print("If you see 502, wait 10–60s and refresh. If 502 persists, paste the last logs and I will debug further.")
else:
    print("\n⚠️ Could not detect trycloudflare URL. Check cloudflared output above for the URL or errors.")
```
